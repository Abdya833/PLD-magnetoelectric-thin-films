# PLD Magnetoelectric Thin Films: Project Data Exploration

This notebook organizes the XRD reflection assignments reported in the supplied project presentation. **It does not reconstruct raw XRD measurements.** The source contains plotted patterns and written assignments, but no numerical 2θ/intensity file.

## 1. Load the documented assignments

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# This path works when the notebook is run from the repository root.
data_path = Path("data/xrd_reflection_assignments.csv")
if not data_path.exists():
    # Also support running the notebook from inside the notebooks/ folder.
    data_path = Path("../data/xrd_reflection_assignments.csv")

assignments = pd.read_csv(data_path)
assignments
```

## 2. Summarize the reflections listed for each material

```python
summary = (
    assignments.groupby("material")["reflection_hkl"]
    .agg(list)
    .reset_index(name="reported_reflections")
)
summary["number_of_listed_reflections"] = summary["reported_reflections"].str.len()
summary
```

## 3. Visualize the number of listed reflection assignments

```python
counts = assignments.groupby("material")["reflection_hkl"].count().sort_index()

ax = counts.plot(kind="bar", figsize=(7, 4))
ax.set_xlabel("Material")
ax.set_ylabel("Number of reflection labels transcribed")
ax.set_title("XRD Reflection Assignments Reported in the Presentation")
ax.grid(axis="y", alpha=0.3)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 4. Interpretation and limitations

The source presentation states that BTO reflections (110), (111), and (200) indicate tetragonal BTO, and CFO reflections (220), (311), and (400) indicate cubic spinel CFO. The bar chart above counts labels transcribed from the presentation; it is **not** a measure of phase fraction, crystallinity, peak intensity, or material quality.

For quantitative analysis, obtain the raw XRD file containing 2θ and intensity, instrument wavelength and scan settings, and suitable reference patterns. Do not infer numerical peak positions from this label-count plot.
